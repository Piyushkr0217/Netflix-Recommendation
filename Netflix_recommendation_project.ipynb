```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
netflix_dataset = pd.read_csv('/content/drive/MyDrive/Netflix dataset/Copy of combined_data_1.txt.zip', header = None,names = ['Cust_ID','Rating'], usecols=[0,1] )
```

```python
netflix_dataset
```

```python
# we can identify the ,movie id by 2 conditions
# 1-1: colon sould be there
# 2- null value will bw there in rating col
```

```python
movie_count = netflix_dataset.isnull().sum()
movie_count = movie_count['Rating']
print('Total movie count : ' , movie_count)
```

```python
total_count = netflix_dataset['Cust_ID'].nunique()
```

```python
customer_count = total_count - movie_count
print('Total customer count : ',customer_count)
```

```python
rating_count = netflix_dataset['Cust_ID'].count() - movie_count
print('Total rating count : ',rating_count)
```

```python
# Stars rated by no. of user

stars = netflix_dataset.groupby('Rating')['Rating'].agg(['count'])
stars
```

```python
ax = stars.plot(kind='barh', legend=False, figsize=(15, 10))
plt.title(f'Total movies: {movie_count}, Total customers: {customer_count}, Total ratings: {rating_count}', fontsize=15, fontweight='bold')
```
